# 2. Nettoyage des données

Plusieurs variables contiennent une modalité `unknown`. On les traite ici, puis on retire deux variables qui ne serviront pas au modèle.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/bank-additional.csv", sep=";")
df_clean = df.copy()

df_clean.shape
```

```python
for col in ["housing", "loan", "marital", "default"]:
    print(df_clean[col].value_counts(), "\n")
```

## Modalités unknown

Pour `housing` et `loan`, on regroupe `unknown` avec `no`. Pour `marital`, il y a très peu de `unknown` (11 clients), on les met avec `single`. Pour `default`, un seul client a `yes`, ce qui est trop peu pour former une catégorie : on le fusionne avec `unknown`.

```python
df_clean["housing"] = df_clean["housing"].replace("unknown", "no")
df_clean["loan"] = df_clean["loan"].replace("unknown", "no")
df_clean["marital"] = df_clean["marital"].replace("unknown", "single")
df_clean["default"] = df_clean["default"].replace("yes", "unknown")

for col in ["housing", "loan", "marital", "default"]:
    print(df_clean[col].value_counts(), "\n")
```

## Variables retirées

`duration` n'est connue qu'une fois l'appel terminé, donc on ne peut pas s'en servir pour prédire. On retire aussi `day_of_week`, qu'on ne garde pas pour la suite.

```python
df_clean = df_clean.drop(columns=["duration", "day_of_week"])

print(df_clean.shape)
print(df_clean.isna().sum().sum(), "valeurs manquantes")
```

```python
# sauvegarde pour l'étape suivante
df_clean.to_csv("../../data/interim/bank_marketing_clean_moussa.csv", index=False, sep=";")
```
