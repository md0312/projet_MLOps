# 1. Data Collection

Cette étape consiste à charger le jeu de données brut et à vérifier qu'il a été correctement importé : dimensions, types des variables et répartition de la cible.

```python
import pandas as pd

# Chargement des données brutes
df = pd.read_csv("../../data/raw/bank-additional.csv", sep=";")

print("Dimensions du jeu de données :", df.shape)
```

```python
# Aperçu des premières lignes
df.head()
```

```python
# Types des variables
print(df.dtypes)
```

```python
# Vérification des valeurs manquantes
print("Nombre de valeurs manquantes :", df.isna().sum().sum())

# Répartition de la variable cible
print("\nRépartition de la cible y :")
print(df["y"].value_counts())
```

### Conclusion

Le jeu de données est correctement chargé. La cible `y` est déséquilibrée : la majorité des clients n'ont pas souscrit. Nous passons maintenant au nettoyage des données.
