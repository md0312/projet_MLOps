# 3. Data Preparation

Cette étape permet de reprendre les données nettoyées et de les organiser pour la suite du projet. Nous vérifions les dimensions et les types de variables, puis nous identifions la cible à prédire avant de passer au Feature Engineering.

```python
import pandas as pd

# Chargement des données nettoyées
df = pd.read_csv("../data/interim/bank_marketing_clean.csv", sep=";")

print("Dimensions du jeu de données :", df.shape)
print("\nTypes des variables :")
print(df.dtypes)
```

Le fichier nettoyé est correctement chargé. Les dimensions et les types des variables sont vérifiés avant de poursuivre la préparation des données.

### 3.1. Identification de la cible et des variables explicatives

La variable `y` indique si le client a souscrit ou non à un dépôt à terme. Nous la séparons des autres variables, qui serviront à expliquer et à prédire cette souscription.

```python
# Séparation de la cible et des variables explicatives
target = "y"

X = df.drop(columns=[target])
y = df[target]

print("Nombre d'observations :", X.shape[0])
print("Nombre de variables explicatives :", X.shape[1])

print("\nRépartition de la cible :")
print(y.value_counts(normalize=True).mul(100).round(2))
```

La cible est désormais séparée des variables explicatives. La répartition de `y` permet de vérifier le déséquilibre entre les clients ayant souscrit et ceux n'ayant pas souscrit, un point à prendre en compte lors de l'évaluation des modèles.

### 3.2. Vérification des variables

Nous vérifions que les variables explicatives et la cible sont correctement séparées et que les données sont prêtes pour la suite du traitement.

```python
print("Dimensions de X :", X.shape)
print("Dimensions de y :", y.shape)

print("\nTypes des variables explicatives :")
print(X.dtypes.value_counts())
```

### Conclusion

Les données nettoyées ont été chargées et la variable cible `y` a été séparée des variables explicatives. Cette étape permet de disposer d'une base organisée pour la suite du projet.
